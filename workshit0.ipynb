```python
def sum_nested_list(nested_list):
    total_sum = 0
    for item in nested_list:
        if isinstance(item, list):
            total_sum += sum_nested_list(item)
        else:
            total_sum += item
    return total_sum

nested_list = [1, [2, [3, 4], 5], 6, [7, 8]]
result = sum_nested_list(nested_list)
print(result)
```

```python
#permutations of characters
def generate_permutations(s):
    if len(s) == 1:
        return [s]
    result = []
    for i in range(len(s)):
        char = s[i]
        remaining = s[:i] + s[i+1:]
        for p in generate_permutations(remaining):
            result.append(char + p)

    return result
print(generate_permutations("abc"))
```

```python
def calculate_directory_size(directory):
    total_size = 0

    for name, value in directory.items():
        if isinstance(value, int):
            total_size += value

        elif isinstance(value, dict):
            total_size += calculate_directory_size(value)

    return total_size

directory_structure = {
    "file1.txt": 200,
    "file2.txt": 300,
    "subdir1": {
        "file3.txt": 400,
        "file4.txt": 100
    },
    "subdir2": {
        "subsubdir1": {
            "file5.txt": 250
        },
        "file6.txt": 150
    }
}

total = calculate_directory_size(directory_structure)
print("Total Size of Directory:", total, "KB")
```

```python
def min_coins_with_combo(coins, amount):
    dp = [float('inf')] * (amount + 1)
    choice = [-1] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for i in range(coin, amount + 1):
            if dp[i - coin] + 1 < dp[i]:
                dp[i] = dp[i - coin] + 1
                choice[i] = coin

    if dp[amount] == float('inf'):
        return -1, []

    combo = []
    x = amount
    while x > 0:
        combo.append(choice[x])
        x -= choice[x]

    return dp[amount], combo

print(min_coins_with_combo([1, 2, 5], 11))
```

```python
def longest_common_subsequence(s1, s2):
    n, m = len(s1), len(s2)
    dp = [[0] * (m + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[n][m]

print(longest_common_subsequence("abcde", "ace"))
```

```python
def knapsack(weights, values, capacity):
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(
                    values[i - 1] + dp[i - 1][w - weights[i - 1]],
                    dp[i - 1][w]
                )
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]

print(knapsack([1, 3, 4, 5], [1, 4, 5, 7], 7))
```
