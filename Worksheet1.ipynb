Matrix Operations:

```python
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

A_inv = np.linalg.inv(A)
print(A.dot(A_inv))
```

```python
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

print(A.dot(B))
print(B.dot(A))
```

```python
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

left = (A.dot(B)).T
right = B.T.dot(A.T)

print(left)
print(right)
```

```python
A = [[ 2, -3,  1],
     [ 1, -1,  2],
     [ 3,  1, -1]]

B = [-1, -3, 9]

solution = np.linalg.solve(A, B)
print(solution)

```
